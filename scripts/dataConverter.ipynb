```python
## Imports and prereqs
import numpy as np
import matplotlib.pyplot as plt
import os
import scipy
import yaml

foldname = 'Day2'
filename = '02062026DataMUO.Txt'

if "data" not in os.getcwd():
    path = os.path.join(os.path.dirname(os.getcwd()), 'data', foldname)
    os.chdir(path)
```

```python
## Load the desired data file and the fitting parameters

def fitFunc(chanNo, params):
    '''Takes in 
    chanNo -- Channel in file
    params -- Parameter and error dictionary
    Returns 2 values: delay, delay error'''
    delayVal = chanNo * params['aval'] + params['bval']
    delayErr = np.sqrt((chanNo * params['aerr'])**2 + (params['berr'])**2)

    return delayVal, delayErr

# Data file
data = np.loadtxt(filename, dtype = float, skiprows=1, delimiter='\t')
head = np.loadtxt(filename, dtype = str,   max_rows=1, delimiter='\t')

# Fitting param file
path = os.path.join(os.path.dirname(os.path.dirname(os.getcwd())), 'data', 'Day2', 'fitparams.yaml')
with open(path, 'r') as f:
    params = yaml.safe_load(f)

```

```python
## Change channels to delay in data file

dvals, derrs = fitFunc(data[:,0], params)
# Change data
data = np.c_[dvals, derrs, data[:,1]]
# Change header
head = 'Delays, Delay errors, ' + head[1]
```

```python
## Save the data

if 'elay' not in filename:
    filename = filename.removesuffix('.Txt') + 'Delay.csv'

np.savetxt(filename, data, header = str(head), delimiter=',')
```
